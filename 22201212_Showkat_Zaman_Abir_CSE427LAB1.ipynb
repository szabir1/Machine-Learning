```python
import numpy as np
```

```python
#1
a = np.random.randint(1, 101, 20)
b = np.random.randint(1, 101, 20)
print("array a:\n", a)
print("array b:\n", b)
```

```python
#2
a = a.reshape(5, 4)
b = b.reshape(5, 4)
print("Matrix a:\n", a)
print("Matrix b:\n", b)
```

```python
#3
scalar_a = a * 4
scalar_b = b * 6
print("Scalar multiplication of matrix a :\n", scalar_a)
print("Scalar multiplication of matrix b :\n", scalar_b)
```

```python
#4
b_mod = b.T
c = np.dot(a, b_mod)
print("Modified Matrix multiplication :\n", c)
```

```python
#5
print("Columns 2 and 3 of matrix c:\n", c[:, 1:3])
```

```python
#6
max_element = np.max(c)
max_index_flat = np.argmax(c)
max_index = np.unravel_index(max_index_flat, c.shape)
print("Largest element in c:", max_element)
print("Index of largest element:", max_index)
```

```python
#7
c_flattened = c.flatten()
print("Flattened of c (1D array):\n", c_flattened)
```

**Matplotlib Library**

```python
import matplotlib.pyplot as plt
```

```python
temperature_dhaka = [25, 34, 21, 45, 28, 6, 43, 18, 7, 2]
humidity_dhaka = [28, 25, 29, 20, 26, 50, 19, 29, 52, 55]

plt.scatter(temperature_dhaka, humidity_dhaka, color='red', marker='*')
plt.title("Temperature vs Humidity (Dhaka)")
plt.xlabel("Temperature (°C)")
plt.ylabel("Humidity (%)")
```

```python
study_hours = [2, 3, 4, 4, 5, 6, 7, 7, 8, 9, 9, 10, 11, 11, 12]
marks_line = [6, 10, 15, 20, 34, 44, 55, 60, 55, 67, 70, 80, 90, 99, 100]
plt.figure(figsize=(12, 8))
plt.plot(study_hours, marks_line, color='b', marker='o', linestyle='dotted', linewidth=2)
plt.title("Study Hours vs Marks")
plt.xlabel("Study Hours")
plt.ylabel("Marks")
```

```python
subjects = ['Maths', 'English', 'Science', 'Physics', 'Computer']
marks = [89, 90, 45, 78, 99]

plt.barh(subjects, marks, color='purple')
plt.title("Scores by Subject (Horizontal)")
plt.xlabel("Marks")
plt.ylabel("Subjects")
```

```python
colors = ['red', 'blue', 'green', 'orange', 'cyan']
plt.bar(subjects, marks, color=colors)
plt.title("Scores by Subject (Vertical)")
plt.xlabel("Subjects")
plt.ylabel("Marks")
```

```python
#5
temperature_dhaka = [25, 34, 21, 45, 28, 6, 43, 18, 7, 2]
humidity_dhaka = [28, 25, 29, 20, 26, 50, 19, 29, 52, 55]
study_hours = [2, 3, 4, 4, 5, 6, 7, 7, 8, 9, 9, 10, 11, 11, 12]
marks = [6, 10, 15, 20, 34, 44, 55, 60, 55, 67, 70, 80, 90, 99, 100]
subjects = ['Maths', 'English', 'Science', 'Physics', 'Computer']
subjects_marks = [89, 90, 45, 78, 99]

plt.figure(figsize=(12, 8))

plt.subplot(2, 2, 1)
plt.scatter(temperature_dhaka, humidity_dhaka, marker='*', color='red')
plt.xlabel('Temperature (°C)')
plt.ylabel('Humidity (%)')
plt.title('Temperature vs Humidity in Dhaka')
plt.grid(True)

# line
plt.subplot(2, 2, 2)
plt.plot(study_hours, marks, marker='o', linestyle='dotted', color='blue')
plt.xlabel('Study Hours')
plt.ylabel('Marks')
plt.title('Study Hours vs Marks')
plt.grid(True)

# Horizontal bar graph
plt.subplot(2, 2, 3)
plt.barh(subjects, subjects_marks, color='purple')
plt.xlabel('Marks')
plt.ylabel('Subjects')
plt.title('Scores in Each Subject (Horizontal Bar)')

# Vertical bar graph
colors = ['red', 'blue', 'green', 'orange', 'cyan']
plt.subplot(2, 2, 4)
plt.bar(subjects, subjects_marks, color=colors)
plt.xlabel('Subjects')
plt.ylabel('Marks')
plt.title('Scores in Each Subject (Vertical Bar)')

plt.tight_layout()
plt.show()
```
